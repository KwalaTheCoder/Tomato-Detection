```python
import cv2
import numpy as np

image_path = "tomato.png"

img = cv2.imread(image_path)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

def nothing(x):
    pass

cv2.namedWindow("HSV Tuner", cv2.WINDOW_NORMAL)
cv2.resizeWindow("HSV Tuner", 900, 700)

# اسلایدرهای min و max برای H، S، V
cv2.createTrackbar("H min", "HSV Tuner", 25, 179, nothing)
cv2.createTrackbar("H max", "HSV Tuner", 65, 179, nothing)
cv2.createTrackbar("S min", "HSV Tuner", 60, 255, nothing)
cv2.createTrackbar("S max", "HSV Tuner", 255, 255, nothing)
cv2.createTrackbar("V min", "HSV Tuner", 60, 255, nothing)
cv2.createTrackbar("V max", "HSV Tuner", 255, 255, nothing)

while True:
    h_min = cv2.getTrackbarPos("H min", "HSV Tuner")
    h_max = cv2.getTrackbarPos("H max", "HSV Tuner")
    s_min = cv2.getTrackbarPos("S min", "HSV Tuner")
    s_max = cv2.getTrackbarPos("S max", "HSV Tuner")
    v_min = cv2.getTrackbarPos("V min", "HSV Tuner")
    v_max = cv2.getTrackbarPos("V max", "HSV Tuner")

    lower = np.array([h_min, s_min, v_min])
    upper = np.array([h_max, s_max, v_max])
    mask = cv2.inRange(hsv, lower, upper)

    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (9, 9))
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel, iterations=3)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel, iterations=2)

    result = cv2.bitwise_and(img, img, mask=mask)

    # نمایش عکس اصلی و ماسک کنار هم
    combined = np.hstack([
        cv2.resize(img, (450, 400)),
        cv2.resize(result, (450, 400))
    ])
    cv2.imshow("HSV Tuner", combined)

    key = cv2.waitKey(1) & 0xFF
    if key == 27:
        break

cv2.destroyAllWindows()
```
